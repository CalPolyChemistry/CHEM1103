## You are the presentation

A slide deck is not the presentation. **You are the presentation.** Your slides should help the audience understand what you are saying, not replace what you are saying.

Your audience should not need to choose between reading paragraphs on the screen and listening to you. In most cases, slides work best when they provide the things that are easier to understand visually: figures, diagrams, images, important numbers, short phrases, and clear take-home messages.

A useful question to ask while designing every slide is:

> **What does this slide help my audience understand while I am speaking?**


# Activity 1 - What makes a good slide?

Your instructor will provide your table with a set of example slides. Work with your group to examine **all eight slides**.

For each slide, discuss:

* What is the slide trying to communicate?
* What works well?
* What makes the slide difficult or ineffective for an audience?
* What would you change?
* Is all of the content worth keeping?

The following list is **not exhaustive**, but these are some things you may want to consider:

* amount of text
* font size and figure-label size
* amount of information on the slide
* use of color and contrast
* accessibility
* figures and tables
* slide titles
* organization and layout
* animation
* whether the audience knows what to look at
* whether the slide supports the speaker or competes with the speaker

Be prepared to discuss any of the eight slides with the class.


# A presentation is a story

The order in which research was **done** is often not the best order in which to **tell the story**.

When you carry out a project, you may collect preliminary data, write code, discover a problem, repeat an experiment, change your analysis, and only later understand which question or result is most important. A presentation does not need to reproduce that chronology.

Instead, think about what your audience needs in order to follow the scientific story.

A useful general structure is:

**Why should we care?** → **What are we asking?** → **How did we answer it?** → **What did we find?** → **What does it mean?** → **What should the audience remember?**

This is a guide, not a required template. Different scientific stories may be organized differently.


# Activity 2 - Build a 10-minute scientific presentation

Your instructor will give your group a set of 16 slide cards from a fictional chemistry project. The cards are numbered in the order the work was done, not in the order they should necessarily appear in a presentation.

The allowed time for the presentation is **10 minutes to present the research, followed by 2 minutes for questions**.

Working together at your table, build the presentation you would give. As you work, decide:

* Which material belongs in the presentation?
* Which material should be removed?
* In what order should the remaining material appear?
* Is anything important missing?
* Should any cards be combined into one slide?
* Which slides need better titles?
* Which slides contain useful information but need to be redesigned?
* Can the presentation you are planning realistically fit into 10 minutes?

There may be more than one reasonable way to organize the presentation. Be prepared to explain the reasoning behind your choices.

# Pacing and delivery

A useful starting point for planning a scientific talk is approximately **one slide per minute**, but this is a guideline rather than a rule.

Some slides may take only 15-30 seconds. A complicated figure or an important result may need considerably longer. The goal is not to make every slide last exactly one minute; the goal is to give the audience enough time to understand the important ideas.

## Guide the audience

When you show a figure, do not assume that the audience instantly sees what you see. Help them understand it:

1. Briefly establish what the figure shows.
2. Direct attention to the important feature.
3. Explain why that feature matters to your scientific story.

Transitions matter too. The audience should understand why one slide follows another.

## Practice out loud

Practice means **giving the presentation out loud while advancing the slides**. Looking through the slides silently is not the same thing.

A timed rehearsal helps you discover:

* explanations that take longer than expected
* awkward transitions
* slides that contain too much information
* slides that are unnecessary
* figures that are difficult to explain
* terminology that is difficult to say clearly
* whether the presentation actually fits the allotted time

Do not solve an overlong presentation by planning to speak faster. **Finishing within the allotted time is part of giving an effective presentation.**


# Giving a group presentation

A group presentation should feel like **one presentation delivered by several people**, not several separate presentations attached together.

Before dividing the speaking responsibilities, agree on the overall scientific story your group wants to tell.

Some useful practices:

* Everyone should understand the whole presentation, not only the slides they personally present.
* Divide the talk into coherent parts of the story rather than simply assigning everyone the same number of slides.
* Work should be equitable, but equitable does not necessarily mean that every person speaks for exactly the same number of seconds.
* Plan transitions between speakers.
* Each speaker should know what comes immediately before and after their section.
* Rehearse together.
* Complete at least one uninterrupted, timed rehearsal with the full group.

Good group work also requires **individual preparation**. A productive planning meeting works better when every group member arrives having already thought about the problem and with ideas to contribute.


# Presentation mechanics and pro tips

Before the presentation:

* Test the presentation file.
* Check that text, figures, labels, equations, and symbols display correctly.
* Make sure figures are readable when projected, not only on your laptop screen.
* Test any animations or videos you plan to use.
* Avoid depending on internet access unless it is necessary.
* Have a backup copy of the presentation.
* Make sure every group member can access the final file.
* Know how you will advance the slides.
* Avoid making major edits immediately before presenting.

Technology should support the presentation rather than becoming part of the presentation.


# Final Activity - Preliminary outline for your project presentation

Your final project presentation will be created and delivered as a group, but this assignment will be completed **individually**.

Before your group develops the final presentation, independently plan the scientific story **you** think your group should tell. Your outline does not need to match the outlines created by your group members. In fact, differences are expected and useful. Good group work means arriving at a planning session having already thought about the problem and having ideas to contribute.

Your project is still in progress, so this is a preliminary plan, not a commitment. Your results may change, and your group may ultimately choose a different organization after comparing everyone's ideas.

Create a preliminary slide-by-slide outline for the presentation. For each proposed slide, include:

1. Slide number
2. Working slide title
3. What the audience will see - for example, a figure, diagram, image, small amount of text, or other visual content
4. The point of the slide - what you want the audience to understand from it

You may use placeholders for work that is not yet complete. For example:

**Slide 6** - How well does our model predict the target property?  
Audience will see: Predicted-versus-actual plot; final version still in progress.  
Point: Show how well the model performs on data it did not use for training and identify any important limitations.

Add as many slides as you think your presentation currently needs. Your outline should reflect **your own current thinking**, even if other members of your group make different choices.

At the end, also answer:

**What part of your proposed presentation are you least certain about right now?**
