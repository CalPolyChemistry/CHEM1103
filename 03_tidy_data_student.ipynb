## What is TIDY data and why use it?

The TIDY data standards are a set of rules used in the data science domain about how data is organized (which is also called "shaped" in data science) that focus on making data machine readable.  The idea is that any code or data analysis tool can work with the data because it adhears to the expected shape.  These rules were first presented in a 2014 paper by Hadley Wickham in the [Journal of Statistical Software](https://www.jstatsoft.org/article/view/v059i10).  

They key principles of TIDY data are:
1. Each variable forms a column.
2. Each observation forms a row.
3. Each cell holds a single measurement.

Entering your data in a spreadsheet does not necessarily make it TIDY.  A spreadsheet is really a graphical interface to your data: it puts your numbers in a grid you can click, sort, and color.  Using spreadsheet tools provides many conveniences. The ability to glance at a table and click a cell is hard to beat while you are collecting data at the bench.

The convenience, unfortunately, often has a cost. The moment you want to do something a spreadsheet was not built for, the data starts to fight you. The solution is to agree on a single, predictable shape for data that every analysis tool “understands”. That shape is tidy data.

In this lesson, you will learn how to record and enter chemical data in spreadsheet programs such that it adhears to TIDY data standards.  We will then discuss "data cleaning", the idea of converting data that is not tidy into data this is.  Finally, we will introduce the python data science library **pandas**, read in tidy data from our spreadsheet, and use it as a data analysis tool.  


### Data shapes

Let's look at a quick example of two "shapes" data can take. Supposed we tracked records of the preparation of sodium chloride solutions that'll be used as a calibration standard for conductivity experiments. If we tracked the **Percent Agreement** (percent agreement = (measured conductivity ÷ expected connectivity) * 100) at three different times, we may decide to use a human-friendly layout that puts the times across the top (as columns):

| Researcher | 8:08 am| 8:18 am | 8:31am |
| -------- | -------- | -------- | ------ |
| A | 97.7% | 101.4% | 95.7% |

This does not conform to tidy data standards.  This so-called **wide** format is easy to eyeball, but it breaks tidy data rule 1: 8:08am, 8:18am, 8:31am are **values** of a "time" variable, not variables themselves. The tidy, or so-called **long**, version gives time its own columns:

| researcher | time_stamp | percent_agreement |
| -------- | -------- | -------- |
| A | 8:08am | 97.7% |
| A | 8:18am | 101.4% |
| A | 8:31am | 95.7% |

It may be harder to read at a glance, but it is much easier for a computer to plot, group, model, or do math on data that is tidily formatted. It's important to remember that we can always wrangle a dataset later to be more human-friendly, but **our core data structure should always be tidy.**  


### An example data set
    
Conductivity is a measure of how well a solution conducts electricity. Pure water conducts electricity very poorly because it contains very few charged particles. When sodium chloride dissolves in water, it separates into sodium (Na⁺) and chloride (Cl⁻) ions. These dissolved ions carry electrical current through the solution, allowing the conductivity of the solution to be measured.

In general, the greater the concentration of dissolved ions, the higher the conductivity. Because the concentration of ions is directly related to how accurately the solution was prepared, conductivity provides a quick, indirect method for evaluating whether the solution is close to the desired concentration.
For this exercise, the expected conductivity of a properly prepared 0.10 M NaCl solution at approximately 25°C is about 10.8 mS/cm. Students compare their measured value to this expected value to evaluate how closely their preparation matches the target solution. A measurement slightly above or below the expected value is completely normal and reflects routine laboratory variation.

Conductivity is commonly reported in either:

  * mS/cm (millisiemens per centimeter) or  
  * µS/cm (microsiemens per centimeter)  

These units measure the same property but differ by a factor of 1000, making them an excellent example of why consistent units are important for data analysis.

Percent Agreement
$$
{\rm Percent Agreement = (Measured Conductivity ÷ Expected Conductivity) × 100 }
$$

A value of 100% would indicate that the measured conductivity exactly matches the expected conductivity of the standard solution. In practice, students should not expect to obtain exactly 100%. Small differences in weighing NaCl, preparing the solution, measuring the final volume, probe calibration, temperature, and normal instrument uncertainty all contribute to variation. Values between approximately 94–104% would be considered reasonable for this instructional laboratory.

### Possible solutions

#### 1. Input values without units 

| Sample Code | Date | Time | NaCl Mass | Volume | Temperature | Conductivity | Percent Agreement |

Units may be understood and assumed. If there is already an established unit convention, you may not need to have the unit in the dataset itself.  One common practices is to have a **data dictionary** as a seperate file from your data, which gives the units and other information about your data.  

However, note how in this exercise that the units are NOT consistent across the provided samples. If there is no unit included (and no conversion calculated), then removing the units makes the dataset very difficult to work with. **Data can be tidy while also being weak in how it is captured.** 

#### 2. Adding units with each value

| Sample Code | Date | Time | NaCl Mass | Volume | Temperature | Conductivity | Percent Agreement |
| -------- | -------- | -------- | -------- | -------- | -------- |-------- | -------- |
| A | 09/14/2026 | 8:08am | 0.584g | 100mL | 22C | 10.55mS/cm | 97.7% |
| B | 09/14/2026 | 8:18am | 582mg | 0.1L | 71.7F | 10950µS/cm | 101.4% |
| C | 09/14/2026 | 8:31am | 0.586g | 100000µL | 295K | 10.34mS/cm | 95.7% |

One way to incorporate units would be to place them directly with the value. This would make sure you do not forget logging the unit in a different column, and would make the individual values more self sufficient. The downside here is the unit information turns the value in the cell into a character/string/text value and you won’t be able to do any numeric calculations without removing the unit from the value. Also, the column will hold 2 bits of information, the value and the unit. In general, you want each cell to only hold 1 bit of information.

#### 3. Adding Unit Info in the Columns 

| Sample Code | Date | Time | NaCl Mass (G) | Volume (mL) | Temperature (C) | Conductivity (mS/cm) | Percent Agreement (%) |
| -------- | -------- | -------- | -------- | -------- | -------- |-------- | -------- |
| A | 09/14/2026 | 8:08am | 0.584 | 100mL | 22.0 | 10.55 | 97.7 |
| B | 09/14/2026 | 8:18am | 0.582 | 100mL | 22.1 | 10.95 | 101.4 |
| C | 09/14/2026 | 8:31am | 0.586 | 100mL | 21.9 | 10.34 | 95.7 |

You can choose to place the unit information in the column names themselves. **Before doing so, you'd need to standardize the values into one consistent unit.** This would make it clearer what the units for a particular value are without having to look up and log a separate column. Chances are that the units will not change across values (after standardization), so it would be highly repetitive to keep the information in a separate column. 

However, be aware of how your column names are formatted. **In general, you want to avoid special characters and spaces in your column names, and if possible keep everything in a single case (e.g., all lower case).** This prevents having errors with leading and trailing spaces (e.g., “sample code” and ” sample code “) when searching for column names, as well as not having to worry about the differences between ”Sample Code”, “sample code”, “sampleCode”, “sample_code”, etc.

#### 4. The Recommended TIDY data Solution

| sample_code | date | time | nacl_mass | volume | temperature | conductivity | percent_agreement |
| -------- | -------- | -------- | -------- | -------- | -------- |-------- | -------- |
| A | 09/14/2026 | 8:08 | 0.584 | 100 | 22.0 | 10.55 | 97.7 |
| B | 09/14/2026 | 8:18 | 0.582 | 100 | 22.1 | 10.95 | 101.4 |
| C | 09/14/2026 | 8:31 | 0.586 | 100 | 21.9 | 10.34 | 95.7 |

This solution isn’t perfect, but it has a lot of features of good spreadsheet data.

  * All the column names are lower-cased and does not contain spaces or special characters
  * Each cell in a column only contains 1 bit of information (separate first and last names, blood pressure values in separate columns, units not part of the value)
  * An (assumed) data dictionary or code book containing the unit information per-column

## The Pandas data science library

[Pandas](https://pandas.pydata.org/docs/) is a Python library used for data analysis and manipulation. Within the world of data science, it is a ubiquitous and widely used library. If you are learning how to analyze data in Python, it will be almost impossible to avoid pandas. 

As with any python library, we need to import pandas to use its functions.  A common practice is to abbreviate `pandas` as `pd`. 

```python
import pandas as pd
```

The central data structure of pandas is called a DataFrame. Pandas DataFrames work very closely with NumPy arrays (Numpy is another important library for working with data) and Pandas dataframes are specifically for data which is two dimensional (rows and columns). NumPy arrays, while similar in some ways, can work with higher dimensional data.  Very broadly, NumPy arrays are designed to work with data that is all one data type, whereas Pandas dataframes are designed to work with a variety of data types.

Pandas is very powerful. In this session, we'll be learning how to access information in pandas dataframes and how to do some basic manipulation and analysis. 

We will use the tidied dataset for the synthesis experiment we dicussed earlier in this lesson.

### Performing calculations with pandas: No more `for` loops!

Both pandas and NumPy dataframes have the convenient feature that they can do element-wise operations and use something called `broadcasting`. This means that if you are doing something like subtracting a number, multiplying, etc to a column or dataframe of information, it can be done all at once instead of with a `for` loop. 

Let's say we wanted to convert the `temp_c` column to temperature in Kelvin.  We can do this and save it as a new column in our dataframe.

```python
list_of_F_temps = [75, 80, 85, 90, 95]

# This does not work correctly with a Python list!
list_of_C_temps = list_of_F_temps + 273.15
print(list_of_C_temps)
```

```python
# This does work with a pandas series!
df['temp_k'] = df['temp_c'] + 273.15

#Show the dataframe again to check our our new columns
df.head()
```

### The `.apply` method

The `.apply` method in pandas is used to apply a function along a row or column of a dataframe.
This is useful when you have a custom function that you need to use on every value in a column, but there is not a NumPy or Pandas function for it.

For example, we could apply the `len` function to our `compound_code` column to get the number of letters in the name for each element.

```python
# Number of letters in name - 
df["compound_code"].apply(len)
```

### Searching and matching data

Frequently in data analysis, you will want to search for particular data in your data set.  The most commonly used function for this operation is `df.loc` which stands for dataframe location.  In the example below, we will search for any batches of our compound where the assay percentage was below 95%.  Then we will print the compound code and the batch ID for these samples, so we know which ones to check.

```python
# Create a dataframe of the matches
matches = df.loc[df["assay_pct"] < 95]

# Print the compound_code and id for the suspect batches
print(matches[["compound_code", "id"]])
```
