## What is data provenance?

**Data provenance is the history of a dataset:** who created it, how and why the observations were obtained, and what happened to them before they reached you. A CSV is one part of that history. Methods, data dictionaries, papers, licenses, and records of processing help you interpret it.

A file that loads successfully is not automatically suitable evidence for a scientific claim. We need to understand what its rows and columns actually represent.

# Trace the provenance of a dataset

| What to investigate | Questions to ask |
| --- | --- |
| Creators and purpose | Who collected or generated the data? Why? Is the uploader also the creator? |
| Observations and methods | What does one row represent? How were values measured, calculated, estimated, or generated? What do the units and labels mean? |
| Coverage | When and where were observations obtained? Which populations, molecules, instruments, or conditions are represented or excluded? |
| Processing | Were files combined, columns renamed, units converted, missing values filled, or observations removed? Can we trace those steps? |
| Evidence and validation | Is there a methods paper, data dictionary, original repository, or validation study? What remains unverified? |
| Rights and credit | What license and restrictions apply to the data? Whom should we cite? Does the data license differ from the article or software license? |
| Version and access | Which file/version are we using? When did we access it? Is there a persistent identifier such as a DOI? |
| Appropriate use | What claims can these data support? What limitations matter for the question we want to answer? |

### Exploring the data with pandas

Before analyzing a dataset, we should examine what it contains, how the observations are distributed, and whether any values are missing.

Two additional useful pandas methods are:

- `value_counts()` counts how many times each unique value appears in a column. For example, penguins["species"].value_counts() tells us how many penguins of each species are represented.

- `isna()` identifies missing values (NaN) in a DataFrame. Combining it with .sum() counts the missing values in each column. For example, penguins.isna().sum() shows how many values are missing from each column.

By default, value_counts() excludes missing values. Adding **dropna=False** includes missing values in the counts.

These commands help us evaluate the completeness and coverage of a dataset without modifying the original data.

```python
# Examine the number of penguins of each species.
display(penguins["species"].value_counts(dropna=False))

# Examine the number of penguins from each island.
display(penguins["island"].value_counts(dropna=False))

# Examine the number of observations from each year.
display(penguins["year"].value_counts(dropna=False))

# Count missing values in each column.
display(penguins.isna().sum())
```

# Activity 2 - Data ethics

Scientists are responsible for how they collect, process, share, and interpret data. A dataset can be easy to analyze and still raise ethical concerns.

| Ethical issue | Key question |
| --- | --- |
| Honesty and transparency | Are origins and values represented accurately? Are measured, predicted, calculated, and generated values distinguished? |
| Bias and representation | Who or what is missing? Might sampling or historical decisions distort a model's conclusions? |
| Privacy and consent | Was collection and sharing appropriate? Could records identify people even after names were removed? |
| Ownership and attribution | Do the terms permit our use? Have we credited original creators and documented modifications? |
| Selection and processing | Are exclusions, missing-data decisions, and transformations justified and recorded? |
| Responsible interpretation | Are conclusions supported by the evidence and its coverage? What consequences could an incorrect prediction have? |

### Data licensing and responsible reuse

Finding a dataset online does not automatically mean that you have permission to use, modify, or redistribute it. **Data licenses specify what others are allowed to do with a dataset.** Before using data, check the license and any restrictions imposed by the original data provider.

Some common licenses include:

| License | What it generally allows | Important requirements |
| --- | --- | --- |
| CC0 | Reuse, modification, and redistribution for almost any purpose | No attribution legally required, although scientific citation is still good practice |
| CC BY | Reuse and modification, including commercial use | Credit the creator, link to the license, and indicate changes |
| CC BY-SA | Reuse and modification, including commercial use | Attribution and share-alike requirements for adaptations |
| CC BY-NC | Reuse and modification for noncommercial purposes | Attribution required; commercial use is restricted |
| CC BY-ND | Sharing unmodified material, including commercially | Attribution required; adapted material cannot be shared under this license |
| ODbL | Reuse and modification of databases | Attribution and database-specific share-alike requirements |
| No license specified | Permissions are unclear | Do not assume that public availability grants permission to reuse or redistribute |

**Important considerations:**

- **Citing data is not the same as having permission to use them.** Attribution does not override licensing restrictions.
- **The uploader may not own the data.** A dataset posted on Kaggle may have originated from another organization with different licensing requirements.
- **Software and data may have different licenses.** Open-source Python code does not necessarily mean the associated dataset is openly licensed.
- **Research and commercial uses may be treated differently.** Some datasets permit educational or academic use but restrict commercial applications.
- **Modified datasets may have additional requirements.** Check whether redistribution or publication of derived datasets is permitted.

For more information, see [Creative Commons licenses](https://creativecommons.org/share-your-work/cclicenses/) and the [Open Data Commons licenses](https://opendatacommons.org/licenses/).

# Activity 3 - Synthetic data: uses and limitations

**Synthetic data are artificially generated records**, rather than direct observations of the events or systems those records describe. They can be produced with random-number generators, statistical models, simulations, or generative AI. A dataset can also mix observed and generated values.

Synthetic data can be useful for teaching, testing software, exploring model assumptions, augmenting limited training data, and studying rare events. They may reduce the need to share sensitive records, but synthetic generation by itself does not guarantee privacy.

| Limitation | What to investigate |
| --- | --- |
| Realism | Which distributions and relationships were reproduced, and which were not? |
| Bias | What source data or assumptions shaped the generator? |
| Artificial patterns | Could the model learn the generator's rules rather than real-world behavior? |
| Validation | Was the intended use checked against relevant independent evidence? |
| Privacy | Was disclosure risk evaluated, including possible reproduction of source records? |
| Documentation | Are generation method, input sources, assumptions, intended uses, and limits explained? |

In chemistry, **calculated properties are not experimental measurements**. For example, the [QM9 dataset](https://doi.org/10.6084/m9.figshare.978904) contains quantum-chemical calculations on molecular structures. These have a documented physical model and computational method, described in [Ramakrishnan et al.](https://www.nature.com/articles/sdata201422). Their accuracy and scope depend on those methods; they should not be presented as measured properties. When evaluating a dataset, you should identify the specific origin of each variable rather than labeling all nonexperimental data the same way. The author of the dataset should document the generation of the data.

## Additional pandas tools for investigating datasets

You've already learned several ways to explore datasets, including examining DataFrames, calculating statistics, and making plots. Here are a few additional commands that may be useful when investigating unfamiliar data.

`sample()` — This method selects random rows from a DataFrame. This is useful because the first few rows may not represent the entire dataset. For example, space.sample(8) displays eight randomly selected rows. You can also write space.sample(8, random_state=1103). The random_state parameter controls the random selection so that you get the same rows each time you run the code. You can use any integer for random_state.

`duplicated()` — This method identifies repeated rows in a DataFrame. Combining it with sum() counts how many rows are duplicates of earlier rows. For example, space.duplicated().sum() counts duplicate records. Remember that duplicate rows aren't necessarily errors; two experimental measurements might legitimately have identical values.

`loc[]` — This method selects rows that satisfy a condition. For example, space.loc[space["Year"] == 2025] selects only missions recorded in 2025. The double equals sign (==) tests whether two values are equal, while a single equals sign (=) assigns a value. You can change 2025 to another year to investigate different observations.

`nunique()` — This method counts how many different values occur in a column. For example, space["Mission Name"].nunique() counts the number of unique mission names. You can replace Mission Name with another column name to investigate other variables. Comparing the number of unique values with the total number of observations can help identify unusual patterns.

**Remember:** A dataset can contain no missing values or duplicates and still be inaccurate or fabricated. Python helps identify unusual patterns, but using your own scientific judgment is essential.

## Optional Bonus - Combining datasets

Scientific data often come in multiple files. We may need to combine these files before analyzing the data. Two common methods are:

* **Concatenation:** Stacks rows from datasets containing similar information, such as measurements collected in different years. The columns should contain comparable information and use consistent units.
* **Merging:** Combines information about the same observations using a shared identifier, such as adding sample concentrations to a table of absorbance measurements.

Both operations can affect the resulting dataset, so it is important to document how the data were combined.

### Merging datasets with pd.merge()

The example below uses two small DataFrames containing absorbance measurements and sample information. These are **invented teaching examples**, not experimental measurements. The same approach can be used with DataFrames loaded from separate files.

The pd.merge() function combines DataFrames using a shared column. The **how** argument determines which observations are retained.

| Join type | What it does |
| --- | --- |
| inner | Keeps only records with matching identifiers in both datasets. |
| left | Keeps all records from the first dataset, adding matching information from the second. |
| right | Keeps all records from the second dataset, adding matching information from the first. |
| outer | Keeps all records from both datasets, using missing values where no match exists. |

A **left join** is often useful for scientific measurements because it preserves the original measurement records, even when some sample information is missing. However, the appropriate join depends on the research question.

For more information, see the [pandas merge documentation](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html).

```python
import pandas as pd

# Create a DataFrame containing absorbance measurements
measurements = pd.DataFrame({
    "sample_id": ["S1", "S1", "S2", "S3"],
    "absorbance": [0.20, 0.22, 0.41, 0.35]
})

# Create a second DataFrame containing sample information
samples = pd.DataFrame({
    "sample_id": ["S1", "S2", "S4"],
    "concentration_mM": [1.0, 2.0, 3.0]
})

# Display the original DataFrames
display(measurements)
display(samples)

# Combine the DataFrames using their shared sample_id column
combined = pd.merge(measurements, samples, on="sample_id", how="left")

# Display the combined DataFrame
display(combined)
```

### What can go wrong when combining datasets?

Combining datasets can introduce errors, even when the Python code runs successfully. Some common problems include:

- **Missing matches:** A sample may appear in one dataset but not the other. For example, S3 has an absorbance measurement but no concentration information. Depending on the join type, its measurement may be retained with missing metadata or excluded entirely.
- **Duplicate identifiers:** If an identifier appears multiple times in both datasets, a merge can create additional rows by matching every possible combination. This can lead to duplicated measurements or incorrect statistical results.
- **Inconsistent identifiers:** Differences in capitalization, spacing, or formatting may prevent records from matching. However, two similar identifiers do not necessarily refer to the same sample.
- **Incompatible data:** Two files might use different units, measurement methods, or definitions for the same variable. Combining them without checking these differences can produce misleading results.
- **Lost provenance:** After combining datasets, it may become difficult to determine where individual values originated unless the original sources and processing steps are documented.

**Good practice:** After merging datasets, check the number of rows, examine missing values, and verify that records were matched correctly. Document the original files, shared identifiers, join type, and any changes made during processing.

Remember that a successful merge does not necessarily mean the resulting dataset is scientifically valid.
