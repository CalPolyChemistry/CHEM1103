### Lists vs. Numpy Arrays

Lists and numpy arrays have many similarities, including the syntax of indexing and slicing. However, numpy arrays have some features that python lists do not. 

````python
# This is a list 
number_list = [1, 2, 3, 5, 7, 11]

# If I wanted to multiple each element by 2, you might think you would do this
number_list2 = number_list * 2

#But this doesn't given what I expected
print(number_list2)
````

If you want to do this using skills you already know, you would have to use a `for` loop, convert each element and save the values to a new list

````python
number_list2 = []
for num in number_list:
    times2 = num*2
    number_list2.append(times2)

print(number_list2)
````

This is actually much easier in numpy and eliminates the need for the `for` loop.  While for our purposes this usually doesn't matter, this would also make your code much faster.

````python
# Recast the list to a numpy array
number_array = np.asarray(number_list)

# Now this does work and makes a new numpy array
number_array2 = number_array*2
print(number_array2)
````

Numpy arrays also have many other features, including the ability to do vector math.  

````python
vector1 = np.asarray([7, 2, 5])
vector2 = np.asarray([1, 1, 3])
difference_vector = vector1 - vector2
print(f'Difference vector: {difference_vector}')
product_vector = vector1*vector2
print(f'Product vector: {product_vector}')
dot_product = np.dot(vector1, vector2)
print(f'The dot product of vector1 and vector2: {dot_product}')
````

## Other looping options

Sometimes you will want to do an operation a certain number of times. 

The general syntax is:

```
for element in list:
  do calculations with element
```

For this you can use a range loop where the syntax is:

```
for index in range(start, end):
  do calculations with index
```

You may also want to combine the power of the for loop you have previously learned with the range loop. You can do this with an enumerate loop using the syntax:

```
for index, element in enumerate(list):
  do calculations with index and/or element from the list
```

Below are some examples.

````python
x = [1, 2, 3, 4, 5]

print('For Loop')
for column in data.columns:
    print(f"The name of column is {column}.")

print('Range Loop')
for i in range(0, data.shape[1]):
    column = data.columns[i]
    print(f"This index {i} corresponds to the column named {column}.")

print('Enumerate Loop')
for i, column in enumerate(data.columns):
    print(f"The value of i is {i}. The value of column is {column}. The mean of data[column] is {data[column].mean()}.")
````

## Creating Arrays for Plotting
NumPy's linspace function allows us to create a range of evenly spaced values over a specified interval. For example, we can create an array of 10 values from 0 to 2π:

`x = np.linspace(0, 2 * np.pi, 10)`

The following example shows you what linspace creates.

## Setting the Figure Size

Most publication figures are smaller than matplotlib creates them by default. This can result in the problem that when you resize a figure like the one above to single-column width (typically 3.3 inches wide), the fonts become too small to read clearly. If you set the size of the figure before you create it, matplotlib will typically make better choices about the relative font size in order to make a publication-quality figure.

You can do this with <code>plt.figure(figsize=(4,3))</code>. Look at the two examples below to see the difference. Note that while the fonts are all the same size, the *relative* size of the font compared to the figure is larger; the second one will look better in a publication.

````python
# Default Figure
plt.plot(t,y,'-')
plt.xlabel('Time (s)')              
plt.ylabel('Vertical Displacement  (cm)')
plt.show()

plt.figure(figsize=(4,3))
plt.plot(t,y,'-')
plt.xlabel('Time (s)')              
plt.ylabel('Vertical Displacement  (cm)')
plt.show()
````

## Plotting Multiple Series on the Same Figure

Adding more than one data series to the same figure is as easy as adding additional plot calls. Looking at the two lines below, both y1 and y2 will be plotted against x on the same figure.

```
plt.plot(x,y1,'-')  # solid line
plt.plot(x,y2,'--') # dashed line
```

Pyplot will choose different colors for you automatically (though you can also tell it what color you want: https://matplotlib.org/stable/users/explain/colors/colors.html), but in most cases it's good practice to use different symbols and not just different colors. You already know how to include different markers, but you can do different line styles as well. Some useful ones to know are `'-'`, `'--'`, `'-.'`, `':'`. Full documentation: https://matplotlib.org/stable/gallery/lines_bars_and_markers/linestyles.html

When plotting multiple data series, you probably want to add a legend. To do this, you can add a label to the plot call and then call `plt.legend()`. See the example below.

````python
epsilon1 = 5e5 # M^-1*cm^-1
epsilon2 = 6e5 # M^-1*cm^-1
b = 1          # cm

c = np.linspace(0,50, 100)  # micromolar

A1 = epsilon1*b*c
A2 = epsilon2*b*c

plt.figure(figsize=(4,3))

plt.plot(c, A1,'-', label='Dye 1') # The label will get applied when the legend is called
plt.plot(c, A2,'--', label='Dye 2')

plt.xlabel('Concentration (mM)') 
plt.ylabel('Absorbance')
plt.legend()

plt.show()
````

matplotlib will choose different colors for you automatically, but you can also specify specific colors by name, e.g. `color='blue'`. If you want to really customize your colors, you can use color hexcodes.  For example, you can look up the official Cal Poly green and gold hexcodes on the [university marketing website](https://ucm.calpoly.edu/color-guidelines).

## Using Greek Letters and Other Math in Labels

Sometimes you need to use Greek letters or other typesetting in your axes labels. You can use MathJax to do so. Anything you put inside dollar signs will be interpreted as math and there are many special characters that you can include.

You may need to include the character r at the start of the label string. Prepending the string with r makes it a "raw string" in Python, which lets you use \ characters without problems.

`plt.ylabel(r'$\Delta H$ $(J\cdot mol^{-1})$')` 

which will render as

$\Delta H$ $(J\cdot mol^{-1})$

A few handy things to know:

- Use Greek letters by using a backslash and their name, e.g. `\delta` for lowercase $\delta$ or `\Delta` for uppercase $\Delta$.
- Superscripts: `x^{-1}` for $x^{-1}$
- Subscripts: `V_{m}` for $V_m$
- Standard state symbol: `\circ` for $\circ$. Combine it with a superscript: `\Delta H^{\circ}` for $\Delta H^{\circ}$

There is a lot more you can do if you're interested: https://math.meta.stackexchange.com/questions/5020/mathjax-basic-tutorial-and-quick-reference

## Using separate y-axes for different series

Sometimes you will want to use two different y-axes because you want to use different scales. If you look at the use of a single axis below, it is hard to see the fluorescence spectrum below.

````python
df = pd.read_csv('data/AlSOAP_spectra.csv')
display(df.head())

plt.figure(figsize=(4,3))
plt.plot(df['wavelength_abs_nm'], df['absorbance'], label='Absorption')
plt.plot(df['wavelength_fl_nm'], df['fluorescence_405_nm'], '--',label='Fluorescence')
plt.xlabel('Wavelength (nm)')
plt.ylabel('Absorbance')
plt.legend()
plt.show()
````

You can instead use a separate y-axis so that the different scales are not problematic.

Notice that there are some differences here when using different axes. In particular, note that when setting labels on individual axes instead of on the entire plot, you must use set_xlabel and set_ylabel instead of xlabel and ylabel.

````python
fig, ax1 = plt.subplots(figsize=(4, 3))

# Left y-axis: absorbance
ax1.plot(
    df['wavelength_abs_nm'],
    df['absorbance'],
    label='Absorption'
)
ax1.set_xlabel('Wavelength (nm)')
ax1.set_ylabel('Absorbance')

# Right y-axis: fluorescence
ax2 = ax1.twinx()

ax2.plot(
    df['wavelength_fl_nm'],
    df['fluorescence_405_nm'],
    '--',
    color='orange',
    label='Fluorescence'
)
ax2.set_ylabel('Fluorescence intensity')

# One combined legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2)

fig.tight_layout()
plt.show()
````

## Multi-Panel Figures

It is common to see multi-panel figures in scientific publications. matplotlib allows for multi-panel layouts.

````python
# Dummy data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create a 2-panel figure (if you want to create a vertical 2-panel figure, the syntax is plt.subplots(2, 1))
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(7, 3))

# --- PANEL 1 (Sine Wave) ---
ax1.plot(x, y1, label=r'$\sin(x)$', color='blue', linewidth=2)
ax1.set_xlabel('Time (s)')
ax1.set_ylabel('Amplitude')

# Add Panel Label 'A' (inside top-left corner)
ax1.text(0.05, 0.95, 'A', transform=ax1.transAxes, fontsize=16, fontweight='bold', va='top')

# --- PANEL 2 (Cosine Wave) ---
ax2.plot(x, y2, label=r'$\cos(x)$', color='red', linewidth=2, linestyle='--')
ax2.set_xlabel('Time (s)')
ax2.set_ylabel('Amplitude')

# Add Panel Label 'B' (inside top-left corner)
ax2.text(0.05, 0.95, 'B', transform=ax2.transAxes, fontsize=16, fontweight='bold', va='top')

# 3. Clean up spacing and render
plt.tight_layout()
plt.show()

````
