## What is the scientific literature?

The term "scientific literature" refers to a broad range of publications that scientists use to communicate with each other. Publication in peer-reviewed journals is an important part of the way in which scientific progress is recorded and allows for the continual advancement of the scientific enterprise. "The literature" is a large and complex system with many different facets; the activities in this lesson are designed to help you understand the basic framework in which scientific communications are published. Over time as a scientist you will learn to effectively utilize and contribute to this important body of knowledge.

## Accessing Subscription Content from Off Campus

Open access publishing has started to become more common, but most publications still require subscription access. We have access to most of the journals with chemistry publications through our library subscription. When you are on campus, you have automatic access to these materials. When you are off campus, you have to take some additional steps to access these materials. There are many methods for this, including through the library's website, but the two efficient methods are listed below.

- **VPN** — You can use the campus virtual private network to make your internet connection act as though you are actually on campus. This is the most robust solution for using campus resources, but also requires the most work.
- **EZProxy** — This is a fast and easy-to-use tool in your web browser that will let you access journals through the library. See the instructions below.

The easiest way to use it is to install the **EZProxy Redirect** browser extension. After you install the extension, right-click the extension icon and select **Options** (this may be different depending on your browser).

![EZProxy settings menu](images/ezproxy-settings-menu.png)

When you land on a journal article page from off campus asking for money, click the **EZProxy Redirect** button in your browser. It will have you authenticate through the portal (if you are not already) and then give you access if we have a subscription.

## Accessing articles by DOI

Most published articles now have a Digital Object Identifier (DOI), which makes it easy to locate them on the web even if they re-organize the website. Some journals are beginning to require the inclusion of DOIs in references sections as well. You can usually find the DOI on the first page of an article. You can find the article by visiting the website doi.org/<DOI>.; For example, for the DOI 10.1021/nl102854s, you would type [doi.org/10.1021/nl102854s](https://doi.org/10.1021/nl102854s) into your browser.
